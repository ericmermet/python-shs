```python
choropleth_map = folium.Map(
    location=[43.6, 1.43],
    tiles="OpenStreetMap",
    zoom_start=10,
)

points = gdf

dfsjoin = gp.sjoin(iris_polys, points) #Spatial join Points to polygons
dfpivot = pd.pivot_table(dfsjoin,index='fid',columns='Generation',aggfunc={'Generation':len})
dfpivot.columns = dfpivot.columns.droplevel()

df_count_by_iris = iris_polys.merge(dfpivot, how='left', on='fid')

#sum men and women in the dataframe
# attention to nan values !!!
df_count_by_iris['All'] = df_count_by_iris[1].fillna(0) + df_count_by_iris[2].fillna(0) + df_count_by_iris[3].fillna(0) + df_count_by_iris[4].fillna(0) + df_count_by_iris[5].fillna(0) + df_count_by_iris[6].fillna(0)

folium.Choropleth(
    geo_data=iris_polys,
    name="choropleth",
    data=df_count_by_iris,
    columns=["fid", "All"],
    key_on="feature.properties.fid",
    fill_color="YlGn",
    fill_opacity=0.7,
    line_opacity=.1,
    legend_name="Choropleth map with pokemon by iris",
).add_to(choropleth_map)

folium.LayerControl().add_to(choropleth_map)

choropleth_map
```

```python
#how to find a good radius for the heat map ?
#analysis of the distance to the nearest neighbour

gdf2 = gdf
gdf.sjoin_nearest(gdf2)
## !!! attention to crs
```

```python
#we need to convert geometry from 4326 (geographical coordinate system) to 2154 (planar coordinate system)
gdf_l93 = gdf.to_crs(2154)
gdf2_l93 = gdf2.to_crs(2154)

gdf_l93.sjoin_nearest(gdf2_l93, distance_col="distances")
#but what about the distances column ?
```

```python
gdf_l93['index_min'] = -1
gdf_l93['distances'] = 0

for index, poi in gdf_l93.iterrows():
    point_a = gdf_l93.loc[index, 'geometry']

    min_distance = 1000000
    min_index = -1

    for index2, poi2 in gdf2_l93.iterrows():
        point_b = gdf2_l93.loc[index2, 'geometry']

        if(point_b != point_a):
            temp_distance = point_a.distance(point_b)
            
            # print('temp_distance ' + str(index) + ' -> ' + str(index2) + ' ' + str(temp_distance) + ' min =' + str(min_distance))
            
            if( temp_distance < min_distance):
                #store index and dist
                min_index = index2
                min_distance = temp_distance
                # print('find a new min ' +  str(index2) + ' ' + str(min_distance))

    gdf_l93['distances'][index] = min_distance
    gdf_l93['index_min'][index] = min_index
    #print( str(index) + ' -> ' + str(min_index) + ' = ' + str(min_distance) )
```

```python
#create a buffer around each point
#2154 is a planar projection for France (in m)
#4326 is a grographic projection, not in meter but in degrees !

gdfbuffer_l93 = gp.GeoDataFrame(gdf2_l93)
gdfbuffer_l93['geometry'] = gdf2_l93.buffer(400, resolution=32) #200 is 200 m

#need to transform planar to geographic for displaying
gdbuffer_4326 = gp.GeoDataFrame(gdfbuffer_l93).to_crs(4326)

gdbuffer_4326

#look at for next:
#https://www.kaggle.com/code/alexisbcook/proximity-analysis
```

```python
# add all polygons to the heatmap
for _, r in gdbuffer_4326.iterrows():
    # Without simplifying the representation of each borough,
    # the map might not be displayed
    sim_geo = gp.GeoSeries(r["geometry"]).simplify(tolerance=0.0001)
    geo_j = sim_geo.to_json()
    geo_j = folium.GeoJson(data=geo_j, style_function=lambda x: {"fillColor": "green"})
    # folium.Popup(r["BoroName"]).add_to(geo_j)
    geo_j.add_to(heat_map)
heat_map
```

```python
#see documentation for voronoi polygon
import matplotlib.pyplot as plt
from geovoronoi import voronoi_regions_from_coords, points_to_coords

coords = points_to_coords(gdf.geometry)

print(coords)

cities = gp.read_file(gp.datasets.get_path('naturalearth_cities'))
#try with naturalearth_lowres

paris = cities[cities.name == 'Paris']
toulouse = cities[cities.name == 'Toulouse']

toulouse_from_iris = iris_polys.dissolve()
#see dissolve documentation  : https://geopandas.org/en/stable/docs/user_guide/aggregation_with_dissolve.html

# toulouse_from_iris.plot()

area_shape = toulouse_from_iris.iloc[0].geometry

region_polys, region_pts = voronoi_regions_from_coords(coords, area_shape)
```

```python
#convert region poly string to a new geodatafram
from shapely import wkt

wkts = []

for poly in region_polys.values():
    # newpolygon = Polygon(poly[0])
    # print(poly)
    wkts.append(poly)

d = {'col1': region_polys.keys(), 'geometry': wkts}

gdf_voronoi = gp.GeoDataFrame(d, crs="EPSG:4326")
```

```python
#and finally map the result
# add all polygons to the heatmap
for _, r in gdf_voronoi.iterrows():
    # Without simplifying the representation of each borough,
    # the map might not be displayed
    sim_geo = gp.GeoSeries(r["geometry"]).simplify(tolerance=0.0001)
    geo_j = sim_geo.to_json()
    geo_j = folium.GeoJson(data=geo_j, style_function=lambda x: {"fillColor": "yellow"})
    # folium.Popup(r["BoroName"]).add_to(geo_j)
    geo_j.add_to(heat_map)
heat_map
```
